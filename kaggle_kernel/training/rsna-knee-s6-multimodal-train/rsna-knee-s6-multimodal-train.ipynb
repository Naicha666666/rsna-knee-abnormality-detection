# 🦴 RSNA Knee: Path B - Full S6 Compact Model Retrain on Multimodal Labels

> **Pipeline Highlights**:
> - **Model Architecture**: ConvNeXt-Small with 12-class spatial attention pooling (`sp_q[c]`).
> - **Training Targets**: 4,349 multimodal refined soft labels (`targets_s6_multimodal.csv`) with 58 gold test holdouts preserved.
> - **Anatomical Anchor Bank v2**: SKM-TEA 3D V-Net cartilage segmentation (`anatomy.pt`) + Patellofemoral 100mm zoom + Baker's cyst local view + Coronal bilateral views.
> - **Laterality Verification**: ResNet-34 side net to prevent medial/lateral mirroring flips.
> - **Loss**: `E_lrprior` logit prior loss with report-state evidence gating.


```python
# ==============================================================================
# Cell 1: Setup Environment and Offline Dependencies
# ==============================================================================
import os, sys, glob, shutil, subprocess, json, time
from pathlib import Path

WORKDIR = Path("/kaggle/working")
INPUTDIR = Path("/kaggle/input")
SRC_DIR = WORKDIR / "src"
SRC_DIR.mkdir(parents=True, exist_ok=True)

print("Starting environment setup...", flush=True)

# 1. Locate assets dataset
def find_dataset(folder_pat):
    for pat in [f"/kaggle/input/{folder_pat}", f"/kaggle/input/*/{folder_pat}", f"/kaggle/input/*/*/{folder_pat}"]:
        matches = sorted(glob.glob(pat))
        if matches:
            return Path(matches[0])
    return None

ortho_assets = find_dataset("rsna-orthofoundation*") or find_dataset("*epoch7*")
assert ortho_assets is not None, "Could not find rsna-orthofoundation dataset in /kaggle/input"
print(f"Found ortho assets at: {ortho_assets}", flush=True)

# 2. Install wheels
wheels_dir = ortho_assets / "dependencies" / "wheels"
if wheels_dir.exists():
    for whl in wheels_dir.glob("*.whl"):
        print(f"Installing wheel: {whl.name}", flush=True)
        subprocess.run([sys.executable, "-m", "pip", "install", "-q", "--no-deps", str(whl)], check=False)

# 3. Setup training module in PYTHONPATH
cloud_training = ortho_assets / "cloud_code" / "training"
if cloud_training.exists():
    target_training = WORKDIR / "training"
    if not target_training.exists():
        shutil.copytree(cloud_training, target_training)
    print(f"Successfully mounted training modules to: {target_training}", flush=True)

sys.path.insert(0, str(WORKDIR))
sys.path.insert(0, str(SRC_DIR))
print("Environment setup complete!", flush=True)
```

```python
# ==============================================================================
# Cell 3: Verify Targets and Cross-Validation Folds
# ==============================================================================
import os, sys, csv, hashlib, glob, shutil
import numpy as np
import pandas as pd
from pathlib import Path

targets_path = Path("/kaggle/working/targets_s6_multimodal.csv")

# 1. Search in /kaggle/input
found_path = None
candidates = [
    Path("/kaggle/input/datasets/easonyanyan/rsna-knee-s6-multimodal-targets/targets_s6_multimodal.csv"),
    Path("/kaggle/input/rsna-knee-s6-multimodal-targets/targets_s6_multimodal.csv"),
    Path("/kaggle/input/targets_s6_multimodal.csv"),
] + [Path(p) for p in sorted(glob.glob("/kaggle/input/**/targets_s6_multimodal.csv", recursive=True))]
for p in candidates:
    if p.exists():
        found_path = p
        break

assert found_path is not None, "targets_s6_multimodal.csv not found in /kaggle/input"
print(f"Found targets dataset at: {found_path}", flush=True)
shutil.copy(found_path, targets_path)

print(f"Targets ready at: {targets_path} ({targets_path.stat().st_size} bytes)", flush=True)

df = pd.read_csv(targets_path)
print(f"Total rows: {len(df)}", flush=True)
gold_count = (df["is_gold"] == 1).sum()
pool_count = (df["is_gold"] == 0).sum()
print(f"Gold test holdouts: {gold_count} (strictly zero training leakage)", flush=True)
print(f"Non-gold training pool: {pool_count}", flush=True)

LABELS = ["ACL", "MCL", "Medial Meniscus", "Lateral Meniscus", "Medial OA", "Lateral OA", "PF OA",
          "Effusion", "Synovitis", "Baker's", "Contusion", "Fracture"]

# Verify hash folds
pool_uids = sorted(df[df["is_gold"] == 0]["StudyInstanceUID"].tolist())
K = 4
folds = [int(hashlib.sha1(u.encode()).hexdigest()[:8], 16) % K for u in pool_uids]
counts = np.bincount(folds, minlength=K).tolist()
print(f"4-Fold Hash Distribution: {counts}", flush=True)
print("Targets and folds verified successfully!", flush=True)
```
